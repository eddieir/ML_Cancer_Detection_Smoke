## TCGA-LUAD / TCGA-LUSC
Requires a GDC token (register at https://portal.gdc.cancer.gov/, "Download Token" from your profile menu) and the `gdc-client` CLI. Set `GDC_TOKEN_FILE` or pass `token_file=` explicitly.

```python
import os
token_file = os.environ.get('GDC_TOKEN_FILE')
if token_file:
    download_all_tcga(token_file)
else:
    print('Set GDC_TOKEN_FILE to a GDC download token to fetch TCGA-LUAD/LUSC.')
```
